```python
## Simple calculator

num1 = float(input("Enter first no."))
num2 = float(input("Enter second no."))
operation = input("Enter operation(+,-, *, /)")

if operation == '+':
    result = num1 + num2
elif operation == '-':
    result = num1 - num2
elif operation == '*':
    result = num1 * num2
elif operation == '/':
    if num2 != 0:
        result = num1 / num2
    else:
        print("Zero division error")
else:
    print("Invalid operation")

print("Result: ", result)
    
```

```python
### Determine the ticket price based on age and whether the person is a student.
# Ticket pricing based on age and student status

age = int(input("enter age "))
is_student = input("Are you student? (yes/no)")

if age<=5:
    price = "Free"

elif age<=12:
    if is_student == "yes":
        price = "10$"
    else:
        price = "12$"

elif age<=17:
    if is_student == "yes":
        price = "15$"
    else:
        price = "20$"
elif age<=64:
    if is_student == "yes":
        price = "25$"
    else:
        price = "30$"
else:
    price = "45$"

print("The price is: ", price)
```

### 2. what is the kind of operation we are doing

```python
[x**2 for x in range(10)]
```

### List comprehension

#### Basic syntax  :-            [expression  for item in iterable]
#### with conditional logic :-   [expression  for item in iterable  if condition]
#### Nested list comprehension   [expression  for item1 in iterable  for item2 in iterable]

```python
### Basic list comprehension

square =  [num**2 for num in range(10)]
print(square)
```

```python
### Without List comprehension with condition
lst=[]
for num in range(10):
    if num%2 == 0:
        lst.append(num)

print(lst)
```

```python
###  List comprehension with condition
[num for num in range(10) if num%2==0]
```

```python
#  Nested list comprehension

lst1 = [1,2,3,4]
lst2 = ['a','b','c','d']

pair = [[i,j] for i in lst1  for j in lst2]
print(pair)
```

```python
### List comprehension with function calls
words = ['hello','world','python','list','bunty']
length = [len(i) for i in words]
print(length)
```
