```python
## Classification Problem

## comments of product is a positive review or negative review
## Reviews ---> eating, eat, eaten => eat(root word)
##              going , gone, goes => go(root word)

words =["eating","eaten","eat","writing","writes","programming","programs","history","finally","finalize"]
```

### __PorterStemmer__

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem("congratulations")
```

```python
stemming.stem("sitting")
```

### __RegexpStemmer Class__
NLTK has RegexpStemmer Class with the help of which we can easily implement Regular expression stemmer algorithms. It basically taks single regular expression and removes any prefix or suffix that matches the expression. Let us see an example|

```python
from nltk.stem import RegexpStemmer
```

```python
reg_Stemmer=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_Stemmer.stem('eating')
```

```python
reg_Stemmer.stem('ingeating')
```

### __Snowball Stemmer__

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstmmr=SnowballStemmer('english')
```

```python
for word in words:
    print(word+"--->"+snowballstmmr.stem(word))
```
